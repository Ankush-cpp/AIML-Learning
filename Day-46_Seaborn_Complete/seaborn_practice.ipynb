```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
tips = sns.load_dataset("tips")

print(tips.head())
print(tips.info())
print(tips.describe())

# 1. Scatter Plot with Hue, Style and Size
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="day",
    style="sex",
    size="size"
)
plt.title("Tips Scatter Plot")
plt.show()

# 2. Line Plot
sns.lineplot(data=tips, x="total_bill", y="tip")
plt.title("Line Plot: Total Bill vs Tip")
plt.show()

# 3. Scatter Plot
sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.title("Scatter Plot")
plt.show()

# 4. Bar Plot
sns.barplot(data=tips, x="day", y="total_bill", hue="sex")
plt.title("Average Total Bill per Day by Sex")
plt.show()

# 5. Box Plot
sns.boxplot(data=tips, x="day", y="total_bill", hue="sex")
plt.title("Total Bill by Day and Sex")
plt.show()

# 6. Violin Plot
sns.violinplot(data=tips, x="day", y="total_bill", hue="sex")
plt.title("Violin Plot of Total Bill")
plt.show()

# 7. Histogram
sns.histplot(data=tips, x="total_bill", bins=10, kde=False)
plt.title("Histogram of Total Bill")
plt.show()

# 8. Histogram with KDE
sns.histplot(data=tips, x="total_bill", bins=10, kde=True)
plt.title("Histogram with KDE")
plt.show()

# 9. Heatmap
numeric_data = tips.select_dtypes(include="number")
corr = numeric_data.corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

# 10. Plot Styling
sns.set_style("whitegrid")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="day"
)
plt.title("Scatter Plot with Whitegrid Style")
plt.show()

# 11. Color Palette
sns.set_palette("pastel")

sns.barplot(data=tips, x="day", y="tip")
plt.title("Bar Plot with Pastel Palette")
plt.show()

# 12. Seaborn with Matplotlib Figure and Axes
fig, ax = plt.subplots(figsize=(7, 5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="day",
    ax=ax
)

ax.set_title("Seaborn with Matplotlib")
ax.set_xlabel("Total Bill")
ax.set_ylabel("Tip")

plt.show()

# 13. Subplots
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    ax=axes[0]
)
axes[0].set_title("Scatter Plot")

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    ax=axes[1]
)
axes[1].set_title("Box Plot")

plt.tight_layout()
plt.show()
```
