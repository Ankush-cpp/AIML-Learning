```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Introduction to Scikit-learn
import sklearn
print(sklearn.__version__)

# Sample Dataset
data = {
    "Hours_Studied": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks": [20, 25, 35, 40, 50, 60, 70, 85]
}

df = pd.DataFrame(data)

print(df.head())

# Features and Target
X = df[["Hours_Studied"]]
y = df["Marks"]

# Import Linear Regression
from sklearn.linear_model import LinearRegression

# Create Model
model = LinearRegression()

# Train Model
model.fit(X, y)

# Best Fit Line Parameters
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)

# Prediction
predictions = model.predict(X)

print("Predictions:", predictions)

# Predict Marks for 9 Hours
new_data = pd.DataFrame({"Hours_Studied": [9]})
result = model.predict(new_data)

print("Predicted Marks for 9 Hours:", result[0])

# Best Fit Line Visualization
plt.scatter(X, y, label="Actual Data")
plt.plot(X, predictions, color="red", label="Best Fit Line")

plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Linear Regression - Best Fit Line")
plt.legend()
plt.show()
```
