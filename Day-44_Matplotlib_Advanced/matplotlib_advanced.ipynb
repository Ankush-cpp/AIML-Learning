```python
import matplotlib.pyplot as plt

# Histogram

data = [12, 15, 14, 18, 20, 22, 19, 25, 30, 45, 16, 21, 24, 28, 32]

plt.hist(data, bins=5)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()


# Multiple Datasets on Histogram

data1 = [12, 15, 14, 18, 20, 22, 19, 25, 30]
data2 = [10, 13, 16, 17, 21, 23, 27, 29, 35]

plt.hist(data1, bins=5, alpha=0.6, label="Dataset 1")
plt.hist(data2, bins=5, alpha=0.6, label="Dataset 2")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Multiple Datasets on Histogram")
plt.legend()
plt.show()


# axvline in Histogram

data = [12, 15, 14, 18, 20, 22, 19, 25, 30, 45, 16, 21, 24, 28, 32]

plt.hist(data, bins=5)
plt.axvline(20, linestyle="--", label="Reference Line")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Histogram with axvline")
plt.legend()
plt.show()


# Box Plot

data = [12, 15, 14, 18, 20, 22, 19, 25, 30, 45]

plt.boxplot(data)
plt.ylabel("Values")
plt.title("Box Plot")
plt.show()


# Operations on Box Plot

data1 = [12, 15, 14, 18, 20, 22, 19, 25]
data2 = [10, 13, 15, 17, 18, 21, 23, 28]
data3 = [8, 12, 14, 16, 20, 24, 26, 35]

plt.boxplot([data1, data2, data3], labels=["Dataset 1", "Dataset 2", "Dataset 3"])
plt.ylabel("Values")
plt.title("Multiple Box Plots")
plt.show()


# Stack Plot

days = [1, 2, 3, 4, 5]
python = [2, 3, 4, 3, 5]
sql = [1, 2, 2, 3, 2]
ml = [1, 1, 2, 2, 3]

plt.stackplot(days, python, sql, ml, labels=["Python", "SQL", "ML"])
plt.xlabel("Days")
plt.ylabel("Hours")
plt.title("Stack Plot")
plt.legend(loc="upper left")
plt.show()


# Subplots

x = [1, 2, 3, 4, 5]
y1 = [10, 20, 15, 25, 30]
y2 = [5, 15, 10, 20, 25]

fig, axes = plt.subplots(1, 2)

axes[0].plot(x, y1)
axes[0].set_title("Line Plot")

axes[1].bar(x, y2)
axes[1].set_title("Bar Plot")

plt.tight_layout()
plt.show()


# Modern Matplotlib

fig, ax = plt.subplots()

ax.plot(x, y1, marker="o", label="Dataset 1")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("Modern Matplotlib")
ax.legend()

plt.show()
```
